### Mask的应用

在注意力机制中，在计算一个查询对应的输出时，需要计算该查询对应每一个键值对的注意力权重

在某些特定的情况下，我们进行一个查询时可能并不想考虑某些特定的键值对

所以通过应用 Mask 来将该查询计算出来的注意力分数中那些不想考虑的键值对的分数置为 -inf

这样的注意力分数通过 softmax 层后，那些分数为 -inf 的键值对最终的权重就是 0（也就是没有考虑该键值对）

<br>

### Padding Mask

对于将一个序列当作键值对的注意力，我们在计算注意力权重时，不需要考虑填充字符

即每次将填充字符对应的注意力分数需要被置为 -inf
